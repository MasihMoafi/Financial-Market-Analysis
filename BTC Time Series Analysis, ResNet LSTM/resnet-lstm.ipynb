```python
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error, mean_absolute_error, accuracy_score

# Load data
btc_data = pd.read_csv('/kaggle/input/btc-daily-data/btc-usd-max.csv',
                       parse_dates=['snapped_at'], index_col='snapped_at')
if not isinstance(btc_data.index, pd.DatetimeIndex):
    btc_data.index = pd.to_datetime(btc_data.index, errors='coerce')
btc_data = btc_data.iloc[250:].sort_index()

# Feature engineering - focusing on returns instead of absolute prices
btc_data['return_1d'] = btc_data['price'].pct_change()
btc_data['volume_change'] = btc_data['total_volume'].pct_change()
btc_data['return_ma5'] = btc_data['return_1d'].rolling(5).mean()
btc_data['return_ma10'] = btc_data['return_1d'].rolling(10).mean()
btc_data['price_std5'] = btc_data['price'].rolling(5).std() / btc_data['price']
btc_data = btc_data.dropna()

# Add target: future returns (what we want to predict)
forecast_horizon = 1
btc_data['target_return'] = btc_data['return_1d'].shift(-forecast_horizon)
btc_data = btc_data.dropna()

# Split data chronologically
train_end = int(len(btc_data) * 0.7)
val_end = int(len(btc_data) * 0.85)

train_data = btc_data.iloc[:train_end]
val_data = btc_data.iloc[train_end:val_end]
test_data = btc_data.iloc[val_end:]

print(f"Train data: {train_data.index[0]} to {train_data.index[-1]}")
print(f"Validation data: {val_data.index[0]} to {val_data.index[-1]}")
print(f"Test data: {test_data.index[0]} to {test_data.index[-1]}")

# Select features, standardize for better training
features = ['return_1d', 'volume_change', 'return_ma5', 'return_ma10', 
            'volume_ma5', 'price_std5']
target = 'target_return'

# Standardize features
scaler = StandardScaler()
train_scaled = scaler.fit_transform(train_data[features])
val_scaled = scaler.transform(val_data[features])
test_scaled = scaler.transform(test_data[features])

# Create sequences
def create_sequences(data, targets, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length + 1):
        X.append(data[i:i+seq_length])
        y.append(targets[i+seq_length-1])
    return np.array(X), np.array(y)

sequence_length = 10
X_train, y_train = create_sequences(train_scaled, train_data[target].values, sequence_length)
X_val, y_val = create_sequences(val_scaled, val_data[target].values, sequence_length)
X_test, y_test = create_sequences(test_scaled, test_data[target].values, sequence_length)

# ResNet+LSTM model (simplified)
def build_model(seq_length, n_features):
    inputs = tf.keras.layers.Input(shape=(seq_length, n_features))
    
    # ResNet block 1
    x = tf.keras.layers.Conv1D(32, 3, padding='same')(inputs)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Activation('relu')(x)
    
    # Residual connection
    skip = tf.keras.layers.Conv1D(32, 1, padding='same')(inputs)
    x = tf.keras.layers.add([x, skip])
    
    # LSTM layers
    x = tf.keras.layers.LSTM(32, return_sequences=True)(x)
    x = tf.keras.layers.Dropout(0.2)(x)
    x = tf.keras.layers.LSTM(16)(x)
    x = tf.keras.layers.Dropout(0.2)(x)
    
    # Output layer - predicting return
    outputs = tf.keras.layers.Dense(1)(x)
    
    model = tf.keras.Model(inputs=inputs, outputs=outputs)
    return model

# Build and compile model
model = build_model(sequence_length, len(features))
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='mse'
)
model.summary()

# Callbacks
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=10, restore_best_weights=True, verbose=1
)
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss', factor=0.5, patience=5, min_lr=0.0001, verbose=1
)

# Train model
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)

# Plot training history
plt.figure(figsize=(12, 6))
plt.plot(history.history['loss'], label='Training loss')
plt.plot(history.history['val_loss'], label='Validation loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid(True)
plt.show()

# Evaluate on test set
y_pred = model.predict(X_test).flatten()

# Calculate metrics (on returns)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("\nTest Set Performance Metrics (on returns):")
print(f"MAE: {mae:.6f}")
print(f"MSE: {mse:.6f}")
print(f"RMSE: {rmse:.6f}")

# Directional accuracy (are we predicting the direction correctly?)
direction_true = (y_test > 0).astype(int)
direction_pred = (y_pred > 0).astype(int)
accuracy = accuracy_score(direction_true, direction_pred)
print(f"Directional Accuracy: {accuracy:.2%}")

# Convert returns to price predictions
test_dates = test_data.index[sequence_length-1:]
actual_prices = test_data['price'].values[sequence_length-1:]
predicted_returns = y_pred
base_prices = test_data['price'].values[sequence_length-1:-forecast_horizon]

# Calculate predicted prices from returns
predicted_prices = base_prices * (1 + predicted_returns[:-forecast_horizon])

# Plot actual vs predicted prices
plt.figure(figsize=(16, 8))
plt.plot(test_dates[:-forecast_horizon], actual_prices[:-forecast_horizon], label='Actual Price', color='blue')
plt.plot(test_dates[:-forecast_horizon], predicted_prices, label='Predicted Price', color='red', linestyle='--')
plt.title('Bitcoin Price vs. Predicted Price (Test Set)')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.show()

# Plot returns
plt.figure(figsize=(16, 6))
plt.plot(test_dates, y_test, label='Actual Returns', color='blue')
plt.plot(test_dates, y_pred, label='Predicted Returns', color='red', linestyle='--')
plt.title('Daily Returns: Actual vs Predicted (Test Set)')
plt.xlabel('Date')
plt.ylabel('Return %')
plt.legend()
plt.grid(True)
plt.axhline(y=0, color='black', linestyle='-')
plt.show()

# Plot cumulative returns
cumulative_actual = np.cumprod(1 + y_test) - 1
cumulative_pred = np.cumprod(1 + y_pred) - 1

plt.figure(figsize=(16, 6))
plt.plot(test_dates, cumulative_actual, label='Actual Cumulative Return', color='blue')
plt.plot(test_dates, cumulative_pred, label='Predicted Cumulative Return', color='red', linestyle='--')
plt.title('Cumulative Returns: Actual vs Predicted (Test Set)')
plt.xlabel('Date')
plt.ylabel('Cumulative Return %')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# The suspiciously good evaluation scores tells me there is something fishy going on! Let's double-check our code for any leakage or overfitting.
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# Load data
btc_data = pd.read_csv('/kaggle/input/btc-daily-data/btc-usd-max.csv',
                       parse_dates=['snapped_at'], index_col='snapped_at')
if not isinstance(btc_data.index, pd.DatetimeIndex):
    btc_data.index = pd.to_datetime(btc_data.index, errors='coerce')
btc_data = btc_data.iloc[250:].sort_index()

# Create features without look-ahead bias
features_df = pd.DataFrame(index=btc_data.index)
features_df['price'] = btc_data['price']
features_df['volume'] = btc_data['total_volume']
features_df['return_1d'] = btc_data['price'].pct_change()
features_df['volume_change'] = btc_data['total_volume'].pct_change()

# Fill only after we have enough data for our features
for i in range(10, len(features_df)):
    price_window = features_df['price'].iloc[i-10:i]
    volume_window = features_df['volume'].iloc[i-10:i]
    return_window = features_df['return_1d'].iloc[i-10:i]
    
    features_df.loc[features_df.index[i], 'ma5'] = price_window[-5:].mean()
    features_df.loc[features_df.index[i], 'ma10'] = price_window.mean()
    features_df.loc[features_df.index[i], 'std5'] = price_window[-5:].std()
    features_df.loc[features_df.index[i], 'vol_ma5'] = volume_window[-5:].mean()
    features_df.loc[features_df.index[i], 'ret_ma3'] = return_window[-3:].mean()

features_df = features_df.dropna()

# Add target - next day's return
features_df['target'] = features_df['return_1d'].shift(-1)
features_df = features_df.dropna()  # Drop rows with NaN targets

# Split train/test - strict time-based split
split_date = features_df.index[-60]  # Use last 60 days as test set
train_data = features_df[features_df.index < split_date].copy()
test_data = features_df[features_df.index >= split_date].copy()

print(f"Training data: {train_data.index[0]} to {train_data.index[-1]}, {len(train_data)} rows")
print(f"Test data: {test_data.index[0]} to {test_data.index[-1]}, {len(test_data)} rows")

# Feature selection
feature_cols = ['return_1d', 'volume_change', 'ma5', 'ma10', 'std5', 'vol_ma5', 'ret_ma3']

# Scale features
scaler = StandardScaler()
X_train = scaler.fit_transform(train_data[feature_cols])
X_test = scaler.transform(test_data[feature_cols])
y_train = train_data['target'].values
y_test = test_data['target'].values

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")

# Build a simpler model to avoid overfitting
model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(len(feature_cols),)),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(1)
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001), loss='mse')

# Train with early stopping
early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history = model.fit(
    X_train, y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=32,
    callbacks=[early_stop],
    verbose=1
)

# Plot training history
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend()
plt.grid(True)
plt.show()

# Predict on test set
test_predictions = model.predict(X_test).flatten()

# Realistic trading simulation
class TradingSimulator:
    def __init__(self, prices, predictions, risk_reward_ratio=2, max_trades=10):
        self.prices = prices
        self.predictions = predictions
        self.risk_reward_ratio = risk_reward_ratio
        self.max_trades = max_trades
        self.starting_capital = 10000
        self.capital = self.starting_capital
        self.position = None
        self.entry_price = None
        self.stop_loss = None
        self.take_profit = None
        self.trades = []
        self.equity_curve = [self.capital]
        self.trade_count = 0
    
    def run_simulation(self, threshold=0.01):
        for i in range(1, len(self.prices)):
            # Update equity curve if not in a position
            if self.position is None:
                self.equity_curve.append(self.capital)
            else:
                current_value = self.capital + self.position * (self.prices[i] - self.entry_price)
                self.equity_curve.append(current_value)
            
            # Check if we need to exit a position
            if self.position is not None:
                # Check stop loss
                if self.prices[i] <= self.stop_loss:
                    loss = (self.stop_loss - self.entry_price) * self.position
                    self.capital += self.position * (self.prices[i] - self.entry_price)
                    self.trades.append({
                        'entry_date': self.entry_date,
                        'exit_date': self.prices.index[i],
                        'entry_price': self.entry_price,
                        'exit_price': self.prices[i],
                        'position': self.position,
                        'pnl': loss,
                        'exit_type': 'Stop Loss'
                    })
                    self.position = None
                    print(f"Exit at stop loss: {self.prices.index[i]}, Price: {self.prices[i]:.2f}, PnL: {loss:.2f}")
                
                # Check take profit
                elif self.prices[i] >= self.take_profit:
                    profit = (self.take_profit - self.entry_price) * self.position
                    self.capital += self.position * (self.prices[i] - self.entry_price)
                    self.trades.append({
                        'entry_date': self.entry_date,
                        'exit_date': self.prices.index[i],
                        'entry_price': self.entry_price,
                        'exit_price': self.prices[i],
                        'position': self.position,
                        'pnl': profit,
                        'exit_type': 'Take Profit'
                    })
                    self.position = None
                    print(f"Exit at take profit: {self.prices.index[i]}, Price: {self.prices[i]:.2f}, PnL: {profit:.2f}")
            
            # Check if we should enter a position
            elif self.trade_count < self.max_trades:
                # Make sure we're not at the last index
                if i < len(self.predictions) and self.predictions[i] > threshold:
                    self.entry_date = self.prices.index[i]
                    self.entry_price = self.prices[i]
                    risk_amount = self.capital * 0.02  # Risk 2% per trade
                    self.stop_loss = self.entry_price * 0.97  # 3% stop loss
                    self.take_profit = self.entry_price * (1 + (0.03 * self.risk_reward_ratio))  # Take profit based on R:R
                    self.position = risk_amount / (self.entry_price - self.stop_loss)
                    self.trade_count += 1
                    print(f"Entry long: {self.entry_date}, Price: {self.entry_price:.2f}, Position: {self.position:.4f}")
                    print(f"SL: {self.stop_loss:.2f}, TP: {self.take_profit:.2f}")
        
        # Close any open position at the end
        if self.position is not None:
            final_price = self.prices[-1]
            pnl = (final_price - self.entry_price) * self.position
            self.capital += pnl
            self.trades.append({
                'entry_date': self.entry_date,
                'exit_date': self.prices.index[-1],
                'entry_price': self.entry_price,
                'exit_price': final_price,
                'position': self.position,
                'pnl': pnl,
                'exit_type': 'End of Period'
            })
            print(f"Exit at end: {self.prices.index[-1]}, Price: {final_price:.2f}, PnL: {pnl:.2f}")
            self.position = None
        
        # Calculate final equity
        self.equity_curve[-1] = self.capital
        
        # Return results
        return {
            'starting_capital': self.starting_capital,
            'ending_capital': self.capital,
            'return': (self.capital / self.starting_capital - 1) * 100,
            'trades': self.trades,
            'equity_curve': self.equity_curve
        }

# Run trading simulation
simulator = TradingSimulator(
    test_data['price'],
    test_predictions,
    risk_reward_ratio=2,
    max_trades=10
)

results = simulator.run_simulation(threshold=0.005)

# Create trade summary
trades_df = pd.DataFrame(results['trades'])
if not trades_df.empty:
    trades_df['return'] = (trades_df['pnl'] / trades_df['entry_price'] * 100)
    
    print("\nTrading Results:")
    print(f"Starting Capital: ${results['starting_capital']:.2f}")
    print(f"Ending Capital: ${results['ending_capital']:.2f}")
    print(f"Total Return: {results['return']:.2f}%")
    print(f"Number of Trades: {len(results['trades'])}")
    
    if len(results['trades']) > 0:
        win_count = len(trades_df[trades_df['pnl'] > 0])
        win_rate = win_count / len(trades_df) * 100
        print(f"Win Rate: {win_rate:.2f}%")
        
        avg_win = trades_df[trades_df['pnl'] > 0]['pnl'].mean() if win_count > 0 else 0
        avg_loss = trades_df[trades_df['pnl'] < 0]['pnl'].mean() if len(trades_df[trades_df['pnl'] < 0]) > 0 else 0
        print(f"Average Win: ${avg_win:.2f}")
        print(f"Average Loss: ${avg_loss:.2f}")
else:
    print("No trades were executed")

# Plot equity curve
plt.figure(figsize=(14, 7))
plt.plot(test_data.index[:len(results['equity_curve'])], results['equity_curve'])
plt.title('Trading Equity Curve')
plt.xlabel('Date')
plt.ylabel('Account Value ($)')
plt.grid(True)
plt.show()

# Plot trades on price chart
plt.figure(figsize=(14, 7))
plt.plot(test_data.index, test_data['price'], label='BTC Price')

# Plot entry and exit points
if not trades_df.empty:
    for _, trade in trades_df.iterrows():
        if trade['pnl'] > 0:
            marker_color = 'green'
        else:
            marker_color = 'red'
        
        plt.scatter(trade['entry_date'], trade['entry_price'], marker='^', color='blue', s=100)
        plt.scatter(trade['exit_date'], trade['exit_price'], marker='v', color=marker_color, s=100)

plt.title('BTC Price with Trade Entry/Exit Points')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.show()

# Model prediction vs actual returns
test_data['predicted_return'] = test_predictions
test_data['next_actual_return'] = test_data['target']

plt.figure(figsize=(14, 7))
plt.scatter(test_data['predicted_return'], test_data['next_actual_return'])
plt.axhline(y=0, color='r', linestyle='-')
plt.axvline(x=0, color='r', linestyle='-')
plt.xlabel('Predicted Return')
plt.ylabel('Actual Return')
plt.title('Prediction Performance: Predicted vs Actual Returns')
plt.grid(True)
plt.show()
```

```python
# Still, the results are suspiciously good. Let's add more rigorous restrictions to avoid any unwarranted data leakage. 

import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from datetime import datetime
import time
import warnings
warnings.filterwarnings('ignore')

# Load data
btc_data = pd.read_csv('/kaggle/input/btc-daily-data/btc-usd-max.csv',
                       parse_dates=['snapped_at'], index_col='snapped_at')
if not isinstance(btc_data.index, pd.DatetimeIndex):
    btc_data.index = pd.to_datetime(btc_data.index, errors='coerce')
btc_data = btc_data.iloc[250:].sort_index()

# Function to create features without look-ahead bias
def create_features(df):
    # Make a copy to avoid modifying the original
    data = df.copy()
    
    # Add basic features
    data['return_1d'] = data['price'].pct_change()
    data['volume_change'] = data['total_volume'].pct_change()
    
    # Use rolling windows explicitly to avoid look-ahead bias
    data['ma5'] = data['price'].rolling(window=5).mean()
    data['ma10'] = data['price'].rolling(window=10).mean()
    data['std5'] = data['price'].rolling(window=5).std()
    data['ret_std5'] = data['return_1d'].rolling(window=5).std()
    data['vol_ma5'] = data['total_volume'].rolling(window=5).mean()
    
    # Drop NaN values that result from rolling windows
    data.dropna(inplace=True)
    
    return data

# Create features
btc_with_features = create_features(btc_data)
print(f"Data ranges from {btc_with_features.index[0]} to {btc_with_features.index[-1]}")

# Implement walk-forward testing with proper time separation
def walk_forward_test(data, train_days=100, test_days=60):
    """
    Walk-forward testing with strict time separation:
    1. Train on a fixed window of past data
    2. Test on the next day only
    3. Move everything forward one day
    4. Repeat for test_days
    """
    # Prepare storage for results
    all_predictions = []
    all_actuals = []
    test_dates = []
    
    # Define feature columns
    feature_cols = ['return_1d', 'volume_change', 'ma5', 'ma10', 'std5', 'ret_std5', 'vol_ma5']
    
    # Make sure we have enough data
    if len(data) < train_days + test_days + 1:
        raise ValueError("Not enough data for the specified training and testing periods")
    
    # Use only the last portion of data to speed up testing
    data = data.iloc[-(train_days + test_days + 10):]
    
    # Loop through each day in the test period
    start_time = time.time()
    print(f"Starting walk-forward testing for {test_days} days...")
    
    for i in range(test_days):
        if i % 10 == 0:  # Print progress every 10 days
            print(f"Processing day {i+1}/{test_days}, elapsed time: {time.time() - start_time:.1f}s")
        
        # Define training and test indices
        train_start = i
        train_end = train_start + train_days
        test_idx = train_end  # Test on the next day after training
        
        # Prepare training data
        train_data = data.iloc[train_start:train_end]
        
        # Prepare test data (just one day)
        test_point = data.iloc[test_idx:test_idx+1]
        
        # Skip if we don't have data for the next day (target)
        if test_idx + 1 >= len(data):
            continue
        
        # Target is next day's return
        train_X = train_data[feature_cols].values
        train_y = train_data['return_1d'].shift(-1).dropna().values
        
        # Remove the last row from training features since we don't have a target for it
        train_X = train_X[:-1]
        
        # Test data
        test_X = test_point[feature_cols].values
        test_date = test_point.index[0]
        
        # The actual next day's return
        if test_idx + 1 < len(data):
            actual_next_return = data.iloc[test_idx + 1]['return_1d']
        else:
            continue  # Skip if we don't have the next day's data
        
        # Scale data based on training set only
        scaler = StandardScaler()
        train_X_scaled = scaler.fit_transform(train_X)
        test_X_scaled = scaler.transform(test_X)
        
        # Simple model for each iteration
        model = tf.keras.Sequential([
            tf.keras.layers.Dense(16, activation='relu', input_shape=(len(feature_cols),)),
            tf.keras.layers.Dropout(0.2),
            tf.keras.layers.Dense(8, activation='relu'),
            tf.keras.layers.Dense(1)
        ])
        
        model.compile(optimizer='adam', loss='mse')
        
        # Train silently
        model.fit(train_X_scaled, train_y, epochs=20, batch_size=16, verbose=0)
        
        # Make prediction
        prediction = model.predict(test_X_scaled, verbose=0)[0][0]
        
        # Store results
        all_predictions.append(prediction)
        all_actuals.append(actual_next_return)
        test_dates.append(test_date)
    
    print(f"Walk-forward testing completed in {time.time() - start_time:.1f}s")
    return pd.DataFrame({
        'date': test_dates,
        'predicted_return': all_predictions,
        'actual_return': all_actuals
    }).set_index('date')

# Run walk-forward testing
results = walk_forward_test(btc_with_features, train_days=100, test_days=60)

# Calculate performance metrics
correlation = results['predicted_return'].corr(results['actual_return'])
mse = ((results['predicted_return'] - results['actual_return']) ** 2).mean()
mae = abs(results['predicted_return'] - results['actual_return']).mean()

print("\nModel Performance Metrics:")
print(f"Correlation: {correlation:.4f}")
print(f"MSE: {mse:.6f}")
print(f"MAE: {mae:.6f}")

# Calculate directional accuracy (if prediction and actual have the same sign)
results['pred_direction'] = results['predicted_return'] > 0
results['actual_direction'] = results['actual_return'] > 0
directional_accuracy = (results['pred_direction'] == results['actual_direction']).mean()
print(f"Directional Accuracy: {directional_accuracy:.2%}")

# Plot predictions vs actuals
plt.figure(figsize=(15, 6))
plt.plot(results.index, results['actual_return'], label='Actual Returns', color='blue')
plt.plot(results.index, results['predicted_return'], label='Predicted Returns', color='red', linestyle='--')
plt.title('Daily Returns: Actual vs Model Predictions')
plt.xlabel('Date')
plt.ylabel('Return %')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot cumulative returns
results['cumulative_actual'] = (1 + results['actual_return']).cumprod() - 1
results['cumulative_pred'] = (1 + results['predicted_return']).cumprod() - 1

plt.figure(figsize=(15, 6))
plt.plot(results.index, results['cumulative_actual'] * 100, label='Actual Cumulative Return', color='blue')
plt.plot(results.index, results['cumulative_pred'] * 100, label='Predicted Cumulative Return', color='red', linestyle='--')
plt.title('Cumulative Returns (%)')
plt.xlabel('Date')
plt.ylabel('Cumulative Return %')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Generate trading signals and simulate trades
def simulate_trading(df, threshold=0.001, risk_reward=2.0, max_trades=10):
    """
    Simulate trading based on return predictions
    - Go long when predicted return > threshold
    - Use a fixed risk-reward ratio for take profit and stop loss
    """
    # Start with a clean dataframe
    trading_df = df.copy()
    
    # Add price data for the same dates
    trading_df['price'] = btc_with_features.loc[trading_df.index]['price']
    
    # Generate signals
    trading_df['signal'] = 0  # 0 = no position, 1 = long
    trading_df.loc[trading_df['predicted_return'] > threshold, 'signal'] = 1
    
    # Initialize trading variables
    capital = 10000
    position = 0
    entry_price = 0
    stop_loss = 0
    take_profit = 0
    trades = []
    equity_curve = [capital]
    trade_count = 0
    in_position = False
    
    # Run simulation
    for i in range(1, len(trading_df)):
        current_date = trading_df.index[i]
        current_price = trading_df['price'].iloc[i]
        current_signal = trading_df['signal'].iloc[i]
        
        # Update equity
        if not in_position:
            equity_curve.append(capital)
        else:
            equity = capital + position * (current_price - entry_price)
            equity_curve.append(equity)
        
        # Check for exit if in position
        if in_position:
            # Check stop loss
            if current_price <= stop_loss:
                capital += position * (stop_loss - entry_price)
                pnl = position * (stop_loss - entry_price)
                trades.append({
                    'entry_date': entry_date,
                    'exit_date': current_date,
                    'entry_price': entry_price,
                    'exit_price': stop_loss,
                    'position': position,
                    'pnl': pnl,
                    'exit_type': 'Stop Loss'
                })
                in_position = False
                print(f"Exit (Stop Loss): {current_date.date()}, Price: ${stop_loss:.2f}, PnL: ${pnl:.2f}")
            
            # Check take profit
            elif current_price >= take_profit:
                capital += position * (take_profit - entry_price)
                pnl = position * (take_profit - entry_price)
                trades.append({
                    'entry_date': entry_date,
                    'exit_date': current_date,
                    'entry_price': entry_price,
                    'exit_price': take_profit,
                    'position': position,
                    'pnl': pnl,
                    'exit_type': 'Take Profit'
                })
                in_position = False
                print(f"Exit (Take Profit): {current_date.date()}, Price: ${take_profit:.2f}, PnL: ${pnl:.2f}")
        
        # Check for entry if not in position and signal is to buy
        elif current_signal == 1 and not in_position and trade_count < max_trades:
            # Calculate position size based on risk
            stop_percent = 0.03  # 3% stop loss
            risk_amount = capital * 0.02  # Risk 2% of capital per trade
            
            entry_price = current_price
            stop_loss = entry_price * (1 - stop_percent)
            take_profit = entry_price * (1 + stop_percent * risk_reward)
            
            # Calculate position size
            position = risk_amount / (entry_price - stop_loss)
            entry_date = current_date
            in_position = True
            trade_count += 1
            
            print(f"Entry: {current_date.date()}, Price: ${entry_price:.2f}, Stop: ${stop_loss:.2f}, Target: ${take_profit:.2f}")
    
    # Close any open position at the end
    if in_position:
        final_price = trading_df['price'].iloc[-1]
        pnl = position * (final_price - entry_price)
        capital += pnl
        trades.append({
            'entry_date': entry_date,
            'exit_date': trading_df.index[-1],
            'entry_price': entry_price,
            'exit_price': final_price,
            'position': position,
            'pnl': pnl,
            'exit_type': 'End of Period'
        })
        print(f"Exit (End of Period): {trading_df.index[-1].date()}, Price: ${final_price:.2f}, PnL: ${pnl:.2f}")
    
    # Calculate results
    trades_df = pd.DataFrame(trades)
    if not trades_df.empty:
        trades_df['return_pct'] = trades_df['pnl'] / (trades_df['entry_price'] * trades_df['position']) * 100
        win_rate = len(trades_df[trades_df['pnl'] > 0]) / len(trades_df) if len(trades_df) > 0 else 0
        
        print("\nTrading Results:")
        print(f"Starting Capital: ${10000:.2f}")
        print(f"Ending Capital: ${capital:.2f}")
        print(f"Total Return: {(capital/10000-1)*100:.2f}%")
        print(f"Number of Trades: {len(trades_df)}")
        print(f"Win Rate: {win_rate:.2%}")
        
        if len(trades_df[trades_df['pnl'] > 0]) > 0:
            print(f"Average Win: ${trades_df[trades_df['pnl'] > 0]['pnl'].mean():.2f}")
        if len(trades_df[trades_df['pnl'] < 0]) > 0:
            print(f"Average Loss: ${trades_df[trades_df['pnl'] < 0]['pnl'].mean():.2f}")
    else:
        print("No trades executed")
    
    return trades_df, equity_curve, trading_df

# Simulate trading
trades_df, equity_curve, trading_df = simulate_trading(results, threshold=0.002, risk_reward=2.0, max_trades=10)

# Plot equity curve
plt.figure(figsize=(15, 6))
plt.plot(results.index[:len(equity_curve)], equity_curve)
plt.title('Trading Equity Curve')
plt.xlabel('Date')
plt.ylabel('Account Value ($)')
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot BTC price with entry/exit points clearly labeled
if not trades_df.empty:
    plt.figure(figsize=(15, 7))
    plt.plot(results.index, trading_df['price'], label='BTC Price')
    
    # Plot entries
    entries_x = trades_df['entry_date']
    entries_y = trades_df['entry_price']
    plt.scatter(entries_x, entries_y, marker='^', color='green', s=100, label='Entry')
    
    # Plot exits with different colors based on exit type
    for idx, trade in trades_df.iterrows():
        if trade['exit_type'] == 'Take Profit':
            color = 'blue'
        elif trade['exit_type'] == 'Stop Loss':
            color = 'red'
        else:
            color = 'black'
        
        plt.scatter(trade['exit_date'], trade['exit_price'], marker='v', color=color, s=100)
        
        # Add connecting line for each trade
        plt.plot([trade['entry_date'], trade['exit_date']], 
                 [trade['entry_price'], trade['exit_price']], 
                 color='gray', linestyle='--', alpha=0.6)
    
    # Add legend for exit types
    plt.scatter([], [], marker='v', color='blue', s=100, label='Take Profit')
    plt.scatter([], [], marker='v', color='red', s=100, label='Stop Loss')
    plt.scatter([], [], marker='v', color='black', s=100, label='End of Period')
    
    plt.title('Bitcoin Price with Trade Entry and Exit Points')
    plt.xlabel('Date')
    plt.ylabel('Price ($)')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
